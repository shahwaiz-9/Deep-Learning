```python
import pandas as pd
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python

# Scaling values
# To avoid biasness

X_train = X_train/255
X_test = X_test/255
```

```python
model = Sequential()
model.add(Flatten(input_shape = (28, 28)))
```

```python
# Making neural network manually

# Input layer
model.add(Dense(64, activation = 'relu'))
model.add(Dense(32, activation='relu'))
#Output layer
model.add(Dense(10, activation = 'softmax')) # Softmax is used for multi classification
```

```python
model.summary()
```

```python
model.compile(loss = 'sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, epochs=10, validation_split = 0.2)
```

```python


pred = model.predict(X_test)
```

```python
y_pred = pred.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_pred, y_test)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 1. Create a grid of points (the 2D space)
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)

# 2. Define two "Lines" (Hidden Neurons)
# Neuron 1: Vertical-ish line (x + 0.5y - 1 = 0)
# Neuron 2: Horizontal-ish line (0.2x + y + 1 = 0)
z1 = 2*X + 1*Y - 2
z2 = -1*X + 2*Y + 2

# Apply Sigmoid Activation to each "line"
h1 = sigmoid(z1)
h2 = sigmoid(z2)

# 3. Combine them in the Output Layer (The Decision)
# Try changing these weights to see the boundary shift!
# Positive weights usually create an "AND" or "OR" intersection.
w1, w2, bias = 5, 5, -7.5
output = sigmoid(w1 * h1 + w2 * h2 + bias)

# --- Plotting ---
fig, ax = plt.subplots(1, 3, figsize=(18, 5))

# Plot Hidden Neuron 1
ax[0].contourf(X, Y, h1, levels=20, cmap='RdBu_r', alpha=0.8)
ax[0].set_title("Hidden Neuron 1 (Line A)")
ax[0].contour(X, Y, z1, levels=[0], colors='black', linestyles='dashed')

# Plot Hidden Neuron 2
ax[1].contourf(X, Y, h2, levels=20, cmap='RdBu_r', alpha=0.8)
ax[1].set_title("Hidden Neuron 2 (Line B)")
ax[1].contour(X, Y, z2, levels=[0], colors='black', linestyles='dashed')

# Plot Combined Output
# The black line here is the final Decision Boundary (where output = 0.5)
ax[2].contourf(X, Y, output, levels=20, cmap='RdBu_r', alpha=0.8)
ax[2].contour(X, Y, output, levels=[0.5], colors='yellow', linewidths=3)
ax[2].set_title("Final Decision Boundary (The Merge)")

for a in ax:
    a.set_xlabel("Input X1")
    a.set_ylabel("Input X2")

plt.tight_layout()
plt.show()
```

In this way we can create any shape ( Decision boundaries using sigmoid in a plane)

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 1. Setup the 2D Input Space
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)

# 2. Raw Linear Lines (Before Sigmoid)
# Neuron 1: 1.5x + 1y - 1 = 0
# Neuron 2: -1x + 1.5y + 1 = 0
z1 = 1.5*X + 1*Y - 1
z2 = -1*X + 1.5*Y + 1

# 3. Hidden Layer Activation (The Squashing)
h1 = sigmoid(z1)
h2 = sigmoid(z2)

# 4. Output Layer Combination (The Merge)
# We weight them and apply one final sigmoid
w1, w2, b_out = 6, 6, -9  # "AND-like" logic
final_z = w1 * h1 + w2 * h2 + b_out
output = sigmoid(final_z)

# --- Visualization ---
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# ROW 1: RAW LINEAR INPUTS
im0 = axes[0, 0].contourf(X, Y, z1, levels=20, cmap='RdGy')
axes[0, 0].set_title("1. Raw Line (Neuron 1)\n$z = w_1x_1 + w_2x_2 + b$")
fig.colorbar(im0, ax=axes[0, 0])

im1 = axes[0, 1].contourf(X, Y, z2, levels=20, cmap='RdGy')
axes[0, 1].set_title("2. Raw Line (Neuron 2)\n$z = w_3x_1 + w_4x_2 + b$")
fig.colorbar(im1, ax=axes[0, 1])

axes[0, 2].axis('off') # Empty space for layout

# ROW 2: ACTIVATED & COMBINED
im2 = axes[1, 0].contourf(X, Y, h1, levels=20, cmap='RdBu_r')
axes[1, 0].set_title("3. Sigmoid Applied (Neuron 1)\n$\sigma(z_1)$")
axes[1, 0].contour(X, Y, h1, levels=[0.5], colors='black', linestyles='--')

im3 = axes[1, 1].contourf(X, Y, h2, levels=20, cmap='RdBu_r')
axes[1, 1].set_title("4. Sigmoid Applied (Neuron 2)\n$\sigma(z_2)$")
axes[1, 1].contour(X, Y, h2, levels=[0.5], colors='black', linestyles='--')

im4 = axes[1, 2].contourf(X, Y, output, levels=20, cmap='RdBu_r')
axes[1, 2].set_title("5. Final Decision Boundary\n$\sigma(\sum w_h h + b_{out})$")
axes[1, 2].contour(X, Y, output, levels=[0.5], colors='yellow', linewidths=3)

plt.tight_layout()
plt.show()
```
