```python
import cv2
from ultralytics import YOLO
from google.colab.patches import cv2_imshow
```

```python
model = YOLO('yolo11n.pt') # Pre Trained Model
```

```python
video_path = "/content/YTDown.com_YouTube_Cars-Moving-On-Road-Stock-Footage-Free-D_Media_Y1jTEyb3wiI_001_1080p.mp4"
cap = cv2.VideoCapture(video_path)
```

```python
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
fps = int(cap.get(cv2.CAP_PROP_FPS))
```

```python
out = cv2.VideoWriter('output_result.mp4', cv2.VideoWriter_fourcc(*'MP4V'), fps, (frame_width, frame_height))

print("Processing video... this may take a minute.")
```

```python
while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    # Run YOLO tracking
    results = model.track(frame, persist=True, verbose=True)

    # Visualize the results on the frame
    annotated_frame = results[0].plot()

    # Write the frame to our new file
    out.write(annotated_frame)
```

```python
cap.release()
out.release()
print("Done! You can now find 'output_result.mp4' in your files tab.")
```

```python
import io
import base64
from IPython.display import HTML

# 1. We need to "re-encode" the video so Chrome/Safari can play it
# This uses 'ffmpeg', a powerful command-line tool pre-installed on Colab
import os
os.system("ffmpeg -i output_result.mp4 -vcodec libx264 -f mp4 output_playable.mp4 -y")

# 2. Function to display the video in the cell
def show_video(file_path):
    video_file = io.open(file_path, 'r+b').read()
    encoded = base64.b64encode(video_file)
    return HTML(data='''<video alt="test" controls width="640">
                <source src="data:video/mp4;base64,{0}" type="video/mp4" />
             </video>'''.format(encoded.decode('ascii')))

# 3. Play it!
show_video('output_playable.mp4')
```
