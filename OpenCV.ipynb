```python
def solution(imagePath):

  input = cv2.imread(imagePath)
  img_rgb = cv2.cvtColor(input, cv2.COLOR_BGR2RGB) # Convert to RGB for display
  gray = cv2.cvtColor(input, cv2.COLOR_BGR2GRAY)

  # Apply Otsu's Thresholding to create a binary mask
  ret, mask = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
  inverted_colors_mask = cv2.bitwise_not(mask)

  # Apply the mask to the original image for result with black background
  otsu_result_black_bg = cv2.bitwise_and(img_rgb, img_rgb, mask=mask)

  # Display results for Otsu's Thresholding
  plt.figure(figsize=(12, 12))
  plt.subplot(2, 2, 1)
  plt.imshow(img_rgb)
  plt.title("Original")
  plt.axis("off")

  plt.subplot(2, 2, 2)
  plt.imshow(gray, cmap="gray")
  plt.title("Grayscale")
  plt.axis("off")

  plt.subplot(2, 2, 3)
  plt.imshow(inverted_colors_mask, cmap='gray')
  plt.title("Otsu's Mask")
  plt.axis('off')

  plt.subplot(2, 2, 4)
  plt.imshow(otsu_result_black_bg)
  plt.title("Otsu's Masked Result (Black BG)")
  plt.axis("off")
  plt.show()

```
