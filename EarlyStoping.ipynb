At some stage the accuracy gets higher then Validation accuracy and the phenomenon is called Overfitting. To tackle this we use method Early stoping.

```python
import matplotlib.pyplot as plt
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers, callbacks
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train.shape
```

```python
X_train = X_train/255
X_test = X_test/255
```

```python
model = Sequential()
model.add(Flatten(input_shape = (28, 28)))
model.add(Dense(128, activation= "relu"))
model.add(Dense(64, activation="relu"))
model.add(Dense(32, activation="relu"))
model.add(Dense(10, activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(loss = "sparse_categorical_crossentropy", optimizer = "Adam", metrics = ["accuracy"])
```

```python
history = model.fit(X_train, y_train, epochs = 100, validation_split = 0.2)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

OPTIMAZING MODEL

```python
model1 = Sequential()
model1.add(Flatten(input_shape = (28, 28)))
model1.add(Dense(128, activation= "relu"))
model1.add(Dense(64, activation="relu"))
model1.add(Dense(32, activation="relu"))
model1.add(Dense(10, activation="softmax"))
```

```python
model1.summary()
```

```python
model.compile(loss = "sparse_categorical_crossentropy", optimizer = "Adam", metrics = ["accuracy"])
```

```python
early_stop = callbacks.EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
```

```python
model_history = model.fit(X_train, y_train, epochs = 100, validation_split=0.2, callbacks=[early_stop])
```

```python
plt.plot(model_history.history['accuracy'])
plt.plot(model_history.history['val_accuracy'])
```

```python
plt.plot(model_history.history['loss'])
plt.plot(model_history.history['val_loss'])
```

Execuation power reduce by 92 iterations. Almost 7 - 8 min training time was reduced.

```python
# much simpler arch model

model2 = Sequential()
model2.add(Flatten(input_shape = (28, 28)))
model2.add(Dense(128, activation= "relu"))
model2.add(Dense(10, activation="softmax"))
```

```python
model2.summary()
```

```python
model2.compile(loss = "sparse_categorical_crossentropy", optimizer = "Adam", metrics = ["accuracy"])
```

```python
model2_history = model2.fit(X_train, y_train, epochs=15, validation_split=0.2)
```

```python
plt.plot(model2_history.history['accuracy'])
plt.plot(model2_history.history['val_accuracy'])
```

```python
plt.plot(model2_history.history['loss'])
plt.plot(model2_history.history['val_loss'])
```
